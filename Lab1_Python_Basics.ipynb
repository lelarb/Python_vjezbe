```python
monthly_consumption = [320, 300, 280, 350, 400, 420]

def total_consumption(consumption):
    total = 0
    for value in consumption:
        total = total + value
    return total

def average_consumption(consumption):
    return total_consumption(consumption) / len(consumption)

def total_cost(consumption, price):
    return total_consumption(consumption) * price

price_per_kwh = 0.15

total = total_consumption(monthly_consumption)
average = average_consumption(monthly_consumption)
cost = total_cost(monthly_consumption, price_per_kwh)

print("Electricity Consumption Report")
print("------------------------------")
print(f"Total consumption: {total} kWh")
print(f"Average consumption: {average:.2f} kWh")
print(f"Total cost: {cost:.2f} EUR")
```
